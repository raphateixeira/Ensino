# Zeros de Funções — Bisseção, Newton e Secante

Notebook complementar à aula [CN02 — Zeros de Funções](../Aulas/CN02ZerosDeFuncoes.qmd).

Implementa as três versões didáticas dos métodos, retornando o **histórico completo** de iterações
$p_n$ e $f(p_n)$ — útil para visualizar a convergência. As versões enxutas usadas nos slides
(que retornam apenas `(raiz, n_iteracoes)`) estão na aula.

```python
import numpy as np
import matplotlib.pyplot as plt
```

## Implementações (com histórico de iterações)

```python
def bissecao_hist(f, a, b, n):
    """Método da bisseção. Retorna (pn, fn) com o histórico das n iterações."""
    pn = np.zeros(n)
    fn = np.zeros(n)
    for i in range(n):
        p = (a + b) / 2
        pn[i], fn[i] = p, f(p)
        if f(a) * f(p) < 0:
            b = p
        else:
            a = p
    return pn, fn
```

```python
def newton_hist(f, df, p0, n):
    """Método de Newton-Raphson. Retorna (pn, fn) com o histórico das n iterações."""
    pn = np.zeros(n)
    fn = np.zeros(n)
    for i in range(n):
        p0 = p0 - f(p0) / df(p0)
        pn[i], fn[i] = p0, f(p0)
    return pn, fn
```

```python
def secante_hist(f, p0, p1, n):
    """Método da secante. Retorna (pn, fn) com o histórico das n iterações."""
    pn = np.zeros(n)
    fn = np.zeros(n)
    for i in range(n):
        p = p1 - f(p1) * (p1 - p0) / (f(p1) - f(p0))
        p0, p1 = p1, p
        pn[i], fn[i] = p, f(p)
    return pn, fn
```

## Exemplo

$$f(x) = \cos(x) - x, \qquad \text{intervalo } [0, 1]$$

Mesmo exemplo usado na aula para comparar a ordem de convergência dos três métodos.

```python
f = lambda x: np.cos(x) - x
df = lambda x: -np.sin(x) - 1

a, b = 0, 1        # intervalo para a bisseção
p0, p1 = 0.3, 0.5  # estimativas iniciais para Newton e secante

N = 7
```

```python
pn1, fn1 = bissecao_hist(f, a, b, N)
pn2, fn2 = newton_hist(f, df, p0, N)
pn3, fn3 = secante_hist(f, p0, p1, N)
```

```python
plt.figure(figsize=(6, 6))
plt.suptitle('Convergência — Bisseção × Newton × Secante')

plt.subplot(211)
plt.plot(pn1, 'o-')
plt.plot(pn2, 'o-')
plt.plot(pn3, 'o-')
plt.ylabel('$p_n$')
plt.grid()
plt.legend(['Bisseção', 'Newton', 'Secante'])

plt.subplot(212)
plt.plot(fn1, 'o-')
plt.plot(fn2, 'o-')
plt.plot(fn3, 'o-')
plt.xlabel('iteração $n$')
plt.ylabel('$f(p_n)$')
plt.grid()

plt.tight_layout()
plt.show()
```

```python
print(f'Bisseção: p_{N} = {pn1[-1]:.10f}')
print(f'Newton:   p_{N} = {pn2[-1]:.10f}')
print(f'Secante:  p_{N} = {pn3[-1]:.10f}')
```

## Comparação com o SciPy

`scipy.optimize.newton` implementa Newton (com `fprime`) ou a secante (sem `fprime`).

```python
from scipy.optimize import newton

p_scipy = newton(f, x0=p0, fprime=df, maxiter=N)
print(f'SciPy (Newton):     {p_scipy:.10f}')
print(f'Implementação própria: {pn2[-1]:.10f}')
```
