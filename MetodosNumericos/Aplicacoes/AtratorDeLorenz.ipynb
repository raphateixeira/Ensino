# Atrator de Lorenz — Caos e Sensibilidade a Condições Iniciais

Notebook complementar à aula [CN08 — EDOs, Sistemas e Estabilidade](../Aulas/CN08EDOsSistemas.qmd).

O sistema de Lorenz é um sistema **não-linear** de 3 EDOs de 1ª ordem, originalmente proposto como
modelo simplificado de convecção atmosférica. É o exemplo clássico de **caos determinístico**:
trajetórias com condições iniciais quase idênticas divergem exponencialmente, embora a dinâmica
seja inteiramente determinística (sem nenhum termo aleatório).

$$\begin{cases}
\dot{x} = \sigma(y - x) \\
\dot{y} = x(\rho - z) - y \\
\dot{z} = xy - \beta z
\end{cases}$$

::: {.callout-note}
Contraste com a análise de estabilidade **linear** da aula CN08 ($y' = \lambda y$): lá, a estabilidade
depende apenas do sinal de $\mathrm{Re}(\lambda)$. Aqui, o sistema é limitado (não diverge para o
infinito) mas ainda assim imprevisível a longo prazo — instabilidade **local** (sensibilidade às
condições iniciais) coexistindo com um atrator **globalmente limitado**.
:::

```python
import numpy as np
import matplotlib.pyplot as plt
```

## Sistema de Lorenz

```python
sigma = 10.0
rho = 28.0
beta = 8 / 3

def lorenz(x):
    """x = [x1, x2, x3] = [x, y, z]"""
    x1, x2, x3 = x
    dx1 = sigma * (x2 - x1)
    dx2 = x1 * (rho - x3) - x2
    dx3 = x1 * x2 - beta * x3
    return np.array([dx1, dx2, dx3])
```

## Duas trajetórias com condições iniciais quase idênticas

$\mathbf{x}_a(0) = [2, -3, 1]$ e $\mathbf{x}_b(0) = [2, -3, 1{,}000001]$ — diferença de $10^{-6}$
apenas em $z(0)$. Integração por Euler explícito (passo fixo $T_s$ pequeno).

```python
t = np.linspace(0, 50, 20000)
Ts = t[1] - t[0]

Xa = np.zeros((len(t), 3))
Xb = np.zeros((len(t), 3))
Xa[0] = [2, -3, 1]
Xb[0] = [2, -3, 1.000001]

for n in range(len(t) - 1):
    Xa[n + 1] = Xa[n] + Ts * lorenz(Xa[n])
    Xb[n + 1] = Xb[n] + Ts * lorenz(Xb[n])
```

## Divergência das trajetórias no tempo

```python
plt.figure(figsize=(7, 5))
labels = ['x', 'y', 'z']
for i in range(3):
    plt.subplot(3, 1, i + 1)
    plt.plot(t, Xa[:, i], label='condição inicial a')
    plt.plot(t, Xb[:, i], label='condição inicial b', alpha=0.7)
    plt.ylabel(labels[i])
    plt.grid()
    if i == 0:
        plt.legend()
plt.xlabel('t')
plt.tight_layout()
plt.show()
```

As duas trajetórias são indistinguíveis por um bom tempo, mas divergem completamente após
algumas dezenas de unidades de tempo — a diferença inicial de $10^{-6}$ é amplificada
exponencialmente (expoente de Lyapunov positivo).

## Retrato de fase e o atrator

```python
plt.figure(figsize=(5, 5))
plt.plot(Xa[:, 0], Xa[:, 1], lw=0.5)
plt.xlabel('x')
plt.ylabel('y')
plt.title('Projeção (x, y)')
plt.grid()
plt.show()
```

```python
plt.figure(figsize=(5, 5))
plt.plot(Xa[:, 1], Xa[:, 2], lw=0.5)
plt.xlabel('y')
plt.ylabel('z')
plt.title('Projeção (y, z)')
plt.grid()
plt.show()
```

```python
fig = plt.figure(figsize=(6, 6))
ax = fig.add_subplot(projection='3d')
ax.plot(Xa[:, 0], Xa[:, 1], Xa[:, 2], lw=0.5, color='C0')
ax.set_xlabel('x')
ax.set_ylabel('y')
ax.set_zlabel('z')
ax.set_title('Atrator de Lorenz')
plt.show()
```

::: {.callout-tip}
Apesar da sensibilidade às condições iniciais, a trajetória permanece confinada a uma região
limitada do espaço de estados — o **atrator estranho**, com sua característica forma de "borboleta".
Isso ilustra por que previsão de longo prazo é impossível em sistemas caóticos (ex.: clima), mesmo
sendo o modelo inteiramente determinístico e sem nenhuma fonte de ruído.
:::
